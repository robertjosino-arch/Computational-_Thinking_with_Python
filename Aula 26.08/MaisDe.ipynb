```python
#Passo 1 — Criar a classe

class Retangulo:
    def __init__(self, nome, base, altura):
        self.nome = nome
        self.base = base
        self.altura = altura

    def area(self):
        return self.base * self.altura

    def perimetro(self):
        return 2 * (self.base + self.altura)

#Passo 2 — Criar os 5 objetos
    
a1 = Retangulo("Retangulo 1", 4, 6)
a2 = Retangulo("Retangulo 2", 3, 8)
a3 = Retangulo("Retangulo 3", 5, 5)
a4 = Retangulo("Retangulo 4", 7, 2)
a5 = Retangulo("Retangulo 5", 10, 3)

#Passo 3 — Colocar os 5 objetos numa lista

lista_retangulo = [a1, a2, a3, a4, a5]

#Passo 4 — Usar o for para montar o dicionário

dados = []

for i in lista_retangulo:
    linha = {
        "Retangulo": i.nome,
        "Base": i.base,
        "Altura": i.altura,
        "Area": i.area(),
        "Perimetro": i.perimetro(),
    }
    dados.append(linha)

print(dados)

#Passo 5 (opcional, se quiser virar tabela como no exemplo)

import pandas as pd
df = pd.DataFrame(dados)
print(df)
```
