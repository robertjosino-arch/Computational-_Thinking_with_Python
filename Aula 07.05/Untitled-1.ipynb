```python
def calculo(x, y):

    soma = x + y
    sub = x - y
    div = x / y
    mult = x * y
    media = (x + y) / 2

    return soma, sub, div, mult, media

print(calculo(100, 10))
```

```python
a, b, c, d, e = calculo(100,10)

print(a)
print(b)
print(c)
print(d)
print(e)
```

```python
# 1. Crie uma função para transformar temperatura Farh em Celsius.

def fahrenheit_para_celsius(e):

    celsius = (e - 32) * 5 / 9

    return celsius


resultado = fahrenheit_para_celsius(100)

print(resultado)
```
