```python
import pandas as pd
import numpy as np

vendas = pd.DataFrame({
    "pedido": [1001, 1002, 1003, 1004, 1005, 1006],
    "produto": ["Notebook", "Mouse", "Monitor", "Teclado", "Notebook", "Webcam"],
    "categoria": ["Informatica", "Acessorio", "Informatica", "Acessorio", "Informatica", "Acessorio"],
    "regiao": ["SP", "RJ", "SP", "MG", "RJ", "SP"],
    "preco": [4200.0, 89.9, 1350.0, 210.0, 3990.0, 149.9],
    "qtd": [2, 10, 3, 5, 1, 4],
})

vendas["Total"] = vendas["preco"] * vendas["qtd"]

#############################################################################

#print(vendas.shape)
#print(vendas.columns)
#print(vendas.info())
#print(vendas[['produto', 'categoria']])
#print(vendas.head(3))
print(vendas.loc[2, 'produto'])

#############################################################################

#Ex. 3
lista_acima = []

for i in range(len(vendas)):
    if vendas['Total'][i] > 1000:
        lista_acima.append(vendas['Total'][i])

print(lista_acima)

#############################################################################

#Ex. 4
mask = vendas['Total'] > 1000
valores_acima = vendas[mask]

print(valores_acima)

#############################################################################

#Exercícios

#Crie um filtro para trazer apeans informações do RJ.
#Crie um filtro para trazer apenas as informações de produtos de informática.
#Crie um filtro para selecionar valores de Total > 1000 apenas para SP.
#Como criar um filtro utilizando duas variáveis (colunas)?

#############################################################################


# Exercício 1: apenas informações do RJ
filtro_rj = vendas['regiao'] == 'RJ'
vendas_rj = vendas[filtro_rj]
print(vendas_rj)

##############################
# Exercício 2: apenas produtos de informática
filtro_info = vendas['categoria'] == 'Informatica'
vendas_info = vendas[filtro_info]
print(vendas_info)

##############################
# Exercício 3: Total > 1000 apenas para SP
filtro_sp_total = (vendas['regiao'] == 'SP') & (vendas['Total'] > 1000)
vendas_sp_total = vendas[filtro_sp_total]
print(vendas_sp_total)

##############################
# Exercício 4: filtro com duas colunas (exemplo geral)
filtro_duas_colunas = (vendas['categoria'] == 'Informatica') & (vendas['regiao'] == 'SP')
vendas_duas_colunas = vendas[filtro_duas_colunas]
print(vendas_duas_colunas)

#############################################################################

```
