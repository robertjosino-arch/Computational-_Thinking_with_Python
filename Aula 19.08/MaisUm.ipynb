```python
#Ex.2

class Pessoa:

    def __init__(self, nome, idade):

        self.nome = nome

        self.idade = idade

    # Método a ser criado

    def apresentar(self):

        return f"Meu nome é {self.nome} e tenho {self.idade} anos!"

# Criar o objeto

p1 = Pessoa("Ana", 35)

print(p1.apresentar())
```

```python
# Crie uma classe para mostrar uma estrutura com as seguintes variáveis: nome, carro, ano e modelo. Crie um método para mostrar um texto com as quatro variáveis

#Ex.3
class Veiculo:
   def __init__(self, nome, carro, ano, modelo):
       self.nome = nome
       self.carro = carro
       self.ano = ano   
       self.modelo = modelo
   def mostrar(self):
       return f"{self.nome} possui um {self.carro}, modelo {self.modelo}, ano {self.ano}."
# Criar o objeto
v1 = Veiculo("Robert", "Ford Fiesta", 2015, "1.6 automático")
print(v1.mostrar())

```

```python
#Exercício (Ex.4):
#Calcular a área e perímetro de um retângulo. Crie as duas variáveis: base e altura. Crie dois métodos: area e perimetro. Apresente os resultados.

#Ex.4
class Retangulo:
   def __init__(self, base, altura):
       self.base = base
       self.altura = altura
   def area(self):
       return self.base * self.altura
   def perimetro(self):
       return 2 * (self.base + self.altura)
# Criar o objeto
r1 = Retangulo(10, 5)
print(f"Área: {r1.area()}")
print(f"Perímetro: {r1.perimetro()}")
```
