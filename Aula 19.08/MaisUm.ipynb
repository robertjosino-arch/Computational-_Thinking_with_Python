```python
# Crie uma classe para mostrar uma estrutura com as seguintes variáveis: nome, carro, ano e modelo. Crie um método para mostrar um texto com as quatro variáveis

#Ex.3
class Veiculo:
   def __init__(self, nome, carro, ano, modelo):
       self.nome = nome
       self.carro = carro
       self.ano = ano   
       self.modelo = modelo
   def mostrar(self):
       return f"{self.nome} possui um {self.carro}, modelo {self.modelo}, ano {self.ano}."
# Criar o objeto
v1 = Veiculo("Robert", "Ford Fiesta", 2015, "1.6 automático")
print(v1.mostrar())

```

```python
#Exercício (Ex.4):
#Calcular a área e perímetro de um retângulo. Crie as duas variáveis: base e altura. Crie dois métodos: area e perimetro. Apresente os resultados.

#Ex.4
class Retangulo:
   
   def __init__(self, base, altura):
       self.base = base
       self.altura = altura
   def area(self):
       return self.base * self.altura
   def perimetro(self):
       return 2 * (self.base + self.altura)
# Criar o objeto
r1 = Retangulo(10, 5)
print(f"Área: {r1.area()}")
print(f"Perímetro: {r1.perimetro()}")
```

```python
class Pessoa:
   def __init__(self, nome, peso, altura):
       self.nome = nome
       self.peso = peso
       self.altura = altura
   def calcular_imc(self):
       return self.peso / (self.altura ** 2)
   def apresentar(self):
       imc = self.calcular_imc()
       return (f"Nome: {self.nome} | Peso: {self.peso}kg | "
               f"Altura: {self.altura}m | IMC: {imc:.2f}")

# Criando 5 pessoas diferentes
pessoas = [
   Pessoa("Laura", 60, 1.65),
   Pessoa("Paulo", 85, 1.80),
   Pessoa("Ana", 55, 1.60),
   Pessoa("Carlos", 90, 1.75),
   Pessoa("Beatriz", 70, 1.70)
]
# Estrutura de repetição para mostrar os dados de cada pessoa
for pessoa in pessoas:
   print(pessoa.apresentar())
```
