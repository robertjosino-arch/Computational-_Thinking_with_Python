```python
import pandas as pd

dados = {
    "regiao": ["Centro", "Zona Norte", "Zona Sul", "Zona Leste", "Zona Oeste", "Industrial"],
    "pm25": [18, 42, 25, 55, 20, 68],
    "pm10": [35, 70, 45, 90, 40, 110],
    "temperatura": [29, 31, 27, 32, 28, 34],
    "umidade": [60, 48, 65, 45, 62, 40]
}

df = pd.DataFrame(dados)

Exercícios

1. Crie a coluna nivel_pm25 utilizando a função apply utilizando função def e estrutura de condição.
Utilize a estrutura de condição sem lambda e com lambda. Nível_pm25.

se pm25 <= 20 - baixo
se pm25 <= 40 - moderado
else - alto

2. Crie a função avaliar_risco se pm25 > 50 e temperatura > 30 - risco alto. se pm25 > 30 - risco moderado, else risco baixo. Com lambda e sem o lambda.

3. Utilizando a função zip, vamos criar uma condição em função de pm25, se pm25> 40, print regiao - qualidade do ar preocupante, else - qualidade do ar aceitável.


##################################################################################################
Exemplos de estruturas com zip

# Ex.1

nomes = ['ana', 'bruno', 'carla']
notas = [19, 22, 21]

for nome, idade in zip(nomes, notas):
    print(nome, idade)
##################################

# Ex.2
produtos = ['arroz', 'feijão', 'macarrão', 'leite']
precos = [25.90, 8.50, 6.90, 5.20]

for produto, preco in zip(produtos, precos):
    print(f'{produto}: R${preco:2f}')

##################################

# Ex.3
produtos = ['notebook', 'mouse', 'teclado', 'monitor']
precos = [3500, 80, 150, 900]
quantidades = [3, 15, 8, 5]

faturamento_total = 0 + 10500 + 1200 + 1200 + 4500

for produto, preco, qtd in zip(produtos, precos, quantidades):
    faturamento = preco * qtd
    faturamento_total += faturamento
    
    print(f'{produto} R$ {faturamento:.2f}')

print(f"Faturamento total - {faturamento_total}")

# Ex.4
nomes = ['ana', 'bruno', 'carla']
notas = [9, 5, 3]

for aluno, nota in zip(nomes, notas):
    
    if nota >= 7:
        situacao = 'Aprovado'
    elif nota >=5:
        situacao = 'Recuperação'
    else:
        situacao = 'Reprovado'
    
    print(aluno, nota, situacao)


# Ex.5
cidades = ['SP', 'AM', 'CE', 'RJ']
temperaturas = [28, 35, 36, 38]

dados = dict(zip(cidades, temperaturas))

for cidade, temperatura in dados.items():
    if temperatura >= 30:
        print("Alerta de calor", cidade)
        
# Ex.6
equipamentos = ['motor', 'transformador', 'gerador']
temperaturas = [75, 92, 81]

for indice, (equipamento, temp) in enumerate(
        zip(equipamentos, temperaturas), start=1):
    print(f"{indice} - {equipamento}: {temp}ºC")

Exercícios

1. Utilizando as listas meses, vendas_2025 e vendas_2026. Calcule/Mostre: 

a) Mostre as 2 informações para cada mês.
b) Calcule a soma das vendas por mês.
c) Mostre o valor total.
d) A variação mensal das vendas.

meses = ['janeiro', 'fevereiro', 'março', 'abril']
vendas_2025 = [10000, 12000, 15000, 13000]
vendas_2026 = [12000, 11000, 18000, 16000]

```

```python
# a) Mostre as 2 informações para cada mês.

meses = ['janeiro', 'fevereiro', 'março', 'abril']
vendas_2025 = [10000, 12000, 15000, 13000]
vendas_2026 = [12000, 11000, 18000, 16000]

for mes, venda_2025, venda_2026 in zip(meses, vendas_2025, vendas_2026):
    print(f'{mes}: 2025 R$ {venda_2025:.2f} | 2026 R$ {venda_2026:.2f}')
```

```python
# b) Calcule a soma das vendas por mês.
 
meses = ['janeiro', 'fevereiro', 'março', 'abril']
vendas_2025 = [10000, 12000, 15000, 13000]
vendas_2026 = [12000, 11000, 18000, 16000]

for mes, venda_2025, venda_2026 in zip(meses, vendas_2025, vendas_2026):
    soma = venda_2025 + venda_2026
    print(f'{mes}: R$ {soma:.2f}')
```

```python
# c) Mostre o valor total.

meses = ['janeiro', 'fevereiro', 'março', 'abril']
vendas_2025 = [10000, 12000, 15000, 13000]
vendas_2026 = [12000, 11000, 18000, 16000]

total_2025 = 0
total_2026 = 0

for venda_2025, venda_2026 in zip(vendas_2025, vendas_2026):
    total_2025 += venda_2025
    total_2026 += venda_2026

total_geral = total_2025 + total_2026

print(f'Total 2025 - R$ {total_2025:.2f}')
print(f'Total 2026 - R$ {total_2026:.2f}')
print(f'Total geral - R$ {total_geral:.2f}')
```

```python
# d) A variação mensal das vendas.

meses = ['janeiro', 'fevereiro', 'março', 'abril']
vendas_2025 = [10000, 12000, 15000, 13000]
vendas_2026 = [12000, 11000, 18000, 16000]

for mes, venda_2025, venda_2026 in zip(meses, vendas_2025, vendas_2026):
    variacao = venda_2026 - venda_2025
    print(f'{mes}: R$ {variacao:.2f}')
```
