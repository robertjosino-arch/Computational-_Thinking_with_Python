```python
#Exercícios

vendedor = ['ana', 'bruno', 'carlos', 'julia']
vendas = [12000, 8500, 15000, 7000]
meta = [10000, 10000, 12000, 8000]

#Crie uma estrutura em python utilizando zip, append, for e crie uma nova coluna
#chamada Situação

#Se venda >= meta - Meta atingida
#Se venda < meta - Abaixo da meta

import pandas as pd

df = pd.DataFrame({
    'vendedores': ['ana', 'bruno', 'carlos', 'julia'],
    'vendas': [12000, 8500, 15000, 7000],
    'metas': [10000, 10000, 12000, 8000]
})

lista_metas = []

for vendedor, venda, meta in zip(df['vendedores'], df['vendas'], df['metas']):
    if venda >= meta:
        lista_metas.append('Meta Atingida')
    else:
        lista_metas.append('Abaixo da Meta')

df['Status_da_Meta'] = lista_metas

print(df)


##########################################################################

#Exercício 2

#sensor = [S1, S2, S3, S4, S5]
#temperatura = [65, 92, 78, 105, 70]
#vibracao = [2.1, 4.8, 3.0, 5.5, 1.9]

#Crie uma estrutura em python utilizando zip, for e condição para mostrar se a 
#classificação do sistema está critico, atenção e normal em função da seguinte lógica:

#temperatura > 90 e vibração > 4 - critico
#temperatura > 80 e vibração > 3.5 - atenção
#Qualquer outra condição - normal

sensor = ['S1', 'S2', 'S3', 'S4', 'S5']
temperatura = [65, 92, 78, 105, 70]
vibracao = [2.1, 4.8, 3.0, 5.5, 1.9]

for s, temp, vib in zip(sensor, temperatura, vibracao):
    if temp > 90 and vib > 4:
        classificacao = 'Crítico'
    elif temp > 80 and vib > 3.5:
        classificacao = 'Atenção'
    else:
        classificacao = 'Normal'
    print(s, temp, vib, classificacao)
    
    
    
    
    
    
    
```
